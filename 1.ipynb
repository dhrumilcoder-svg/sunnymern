```python
import torch

x = torch.tensor(2.0, requires_grad=True)

  # gradient - best fit line //slop 

#dy/dx = 2x + 3
#x = 2
#gradient = 7
#y=x2+3x+1

y = x**2 + 3*x + 1

y.backward() # forward backward loss fucntion 

print(x.grad)
```
