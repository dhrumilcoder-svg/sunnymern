```python


from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python


model = YOLO("yolov8n.pt")

print("Model Loaded Successfully")
```

```python
import time
cap = cv2.VideoCapture(0)
time.sleep(2) # Give the webcam time to initialize
print("Webcam Started") # time class 
```

```python
ret, frame = cap.read()    # frame - videcap - me thod = 
cap.release()

if ret:
    print("Image Captured Successfully")
else:
    print("Camera Error: Ensure webcam is not being used by another application.")
```

```python
if ret:
    img = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    print("Color conversion successful.")
      # yolo - you only looks once algoritham store - db - challan 
      # what is - set of rules /instruction a
      
else:
    print("Cannot convert an empty frame.")
```

```python
if ret:
    results = model(img)
    output = results[0].plot() # Get annotated image with bounding boxes
    print("Detection Completed")
else:
    print("No image to run detection on.")
```

```python
import matplotlib.pyplot as plt

if ret:
    plt.figure(figsize=(12,8))
    plt.imshow(output)
    plt.title("YOLO Face + Object Detection")
    plt.axis("off")
    plt.show()
else:
    print("No output to display.")
```

```python

names = model.names
detected = results[0].boxes.cls.tolist()

print("Detected Objects:")

for obj in detected:
    print("-", names[int(obj)])
```
