```python
import tensorflow as tf

a = tf.constant([1, 2, 3]) #tf.constant() = never change
print(a)
```

```python
x = tf.Variable([10, 20, 30])
print(x) 
```

```python
a = tf.constant(5)
b = tf.constant(3)

print(tf.add(a, b))   
print(tf.subtract(a, b))
print(tf.multiply(a, b)) 
print(tf.divide(a, b))  
```

```python
list_data = [1,2,3]
tensor_data = tf.convert_to_tensor(list_data)
print(tensor_data) #python list to tensor
```
