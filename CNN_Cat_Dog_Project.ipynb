```python
import tensorflow as tf  # Deep Learning mate
from tensorflow.keras import layers, models  # Model banava mate
from tensorflow.keras.preprocessing.image import ImageDataGenerator  # Image load karva mate
import matplotlib.pyplot as plt  # Graph mate
```

```python
# pipeline - 1 layer cnn 
model = models.Sequential() #pipeline
# 2d - 
# first cnn layer // features - 32 - 3 * 3 mat
model.add(layers.Conv2D(32, (3,3), activation='relu', input_shape=(64,64,3)))
# 32 - find 32 features from the image
#(3,3) means a filter of 3 rows × 3 columns
#It is a small box (window) that moves (slides) over the image
#The filter contains 9 values (weights)
#As it slides across the image, it:
#looks at a small part of the image at a time
# performs calculations
#and detects important features (like edges, patterns, shapes)
#input_shape = (64,64,3)
#64x64 image size
#3 = RGB (color image)
# captch 
model.add(layers.MaxPooling2D((2,2)))
#reduce image size and store importance features 64x64 → 32x32

#Second Convolution Layer
#more complex patterns (eyes, shapes, textures)
model.add(layers.Conv2D(64, (3,3), activation='relu'))

#model fast and efficient
model.add(layers.MaxPooling2D((2,2)))#32 32

# Flatten  2D data → 1D vector
model.add(layers.Flatten())

#Fully Connected Layer
#ANN part starts here:

#It has 64 neurons 32 fe
#It learns to make decisions based on the extracted features
model.add(layers.Dense(64, activation='relu'))


model.add(layers.Dense(1, activation='sigmoid'))

# Fully Connected Layer -dense
# 1 mean output layer have 1 nuerons 

model.summary()
#output shape
```

```python
#  Step 4: Compile
#optimizer algoritham - how to learn model = 'adam'
# 
#Cross-entropy measures how wrong the prediction is
#Performance check -  metrics=['accuracy']
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
#  Step 5: Train
history = model.fit(train_generator, epochs=35, validation_data=test_generator)

#validation_data=test_generator- During training, we check using separate data whether the model has learned correctly or not.
```

```python
#  Step 6: Accuracy Graph
#history.history['accuracy']- each epoch accuracy store
plt.plot(history.history['accuracy'], label='train')


# epoch 
#Real performance wrong predication //30 
plt.plot(history.history['val_accuracy'], label='test')

#display label in graph line
plt.legend()

# display graph in screen
plt.show()
```
